```python
import requests
from bs4 import BeautifulSoup

def get_latest_gencode_fasta(data_dir='./'):
    base_url = "https://ftp.ebi.ac.uk/pub/databases/gencode/Gencode_human/latest_release/"
    response = requests.get(base_url)
    soup = BeautifulSoup(response.text, 'html.parser')
    
    latest_release = [a['href'] for a in soup.find_all('a', href=True) if a['href'].startswith('gencode') and a['href'].endswith('transcripts.fa.gz')][-1]
    fasta_url = f"{base_url}{latest_release}"

    # Download the MD5SUMS file
    fasta_response = requests.get(f"{base_url}MD5SUMS")
    with open(f"{data_dir}MD5SUMS", "wb") as f:
        f.write(fasta_response.content)
        
    print(f"Downloaded: {base_url}MD5SUMS")
    
    # Download the FASTA file
    fasta_response = requests.get(fasta_url)
    with open(f"{data_dir}{latest_release}", "wb") as f:
        f.write(fasta_response.content)

    print(f"Downloaded: {fasta_url}")

# Example usage
get_latest_gencode_fasta()
```
